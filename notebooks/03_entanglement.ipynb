## Exercise 1: Create a Bell State

The Bell state is the simplest entangled state: |Φ+⟩ = (|00⟩ + |11⟩) / √2

```
# Create Bell state
qc = QuantumCircuit(2, 2)

# Step 1: Put qubit 0 in superposition
qc.h(0)

# Step 2: CNOT with qubit 0 as control, qubit 1 as target
qc.cx(0, 1)

# Measure both qubits
qc.measure([0, 1], [0, 1])

qc.draw('mpl')
```

```
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
print("Results:", result.get_counts())
plot_histogram(result.get_counts())
```

**Key Observation:** We only see |00⟩ and |11⟩, never |01⟩ or |10⟩!

The qubits are *perfectly correlated* - they always match.

## Exercise 2: Verify Entanglement vs Classical Correlation

Let's compare entanglement to a classical scenario where we simply copy a random bit.

```
# Look at the quantum state BEFORE measurement
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)

state = Statevector(qc)
print("Quantum state:")
print(state)
print("\nThis is a superposition of |00⟩ and |11⟩ - neither qubit has a definite value!")
```

## Exercise 3: Other Bell States

There are 4 Bell states. Let's create them all.

```
# |Φ+⟩ = (|00⟩ + |11⟩) / √2 - we already made this
# |Φ-⟩ = (|00⟩ - |11⟩) / √2
# |Ψ+⟩ = (|01⟩ + |10⟩) / √2
# |Ψ-⟩ = (|01⟩ - |10⟩) / √2

def create_bell_state(state_type):
    qc = QuantumCircuit(2, 2)
    qc.h(0)
    qc.cx(0, 1)
    
    if state_type == 'phi_minus':
        qc.z(0)  # Add phase
    elif state_type == 'psi_plus':
        qc.x(1)  # Flip qubit 1
    elif state_type == 'psi_minus':
        qc.z(0)
        qc.x(1)
    
    qc.measure([0, 1], [0, 1])
    return qc

# Test each Bell state
for name in ['phi_plus', 'phi_minus', 'psi_plus', 'psi_minus']:
    qc = create_bell_state(name)
    result = simulator.run(qc, shots=1000).result()
    print(f"{name}: {result.get_counts()}")
```

## Exercise 4: Entanglement with 3 Qubits (GHZ State)

The GHZ state extends entanglement to 3 qubits: (|000⟩ + |111⟩) / √2

```
qc = QuantumCircuit(3, 3)

# Create GHZ state
qc.h(0)
qc.cx(0, 1)
qc.cx(0, 2)

qc.measure([0, 1, 2], [0, 1, 2])
qc.draw('mpl')
```

```
result = simulator.run(qc, shots=1000).result()
print("GHZ state results:", result.get_counts())
plot_histogram(result.get_counts())
```

**Observation:** All 3 qubits are now entangled - they're ALL 0 or ALL 1.
