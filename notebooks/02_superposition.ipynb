## Exercise 1: Multiple Qubits in Superposition

With n qubits in superposition, we represent 2^n states simultaneously.

```
# 2 qubits = 4 possible states (00, 01, 10, 11)
qc = QuantumCircuit(2, 2)
qc.h(0)  # Put qubit 0 in superposition
qc.h(1)  # Put qubit 1 in superposition
qc.measure([0, 1], [0, 1])

qc.draw('mpl')
```

```
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
print("Results:", result.get_counts())
plot_histogram(result.get_counts())
```

**Observation:** Each of the 4 states appears with roughly 25% probability.

## Exercise 2: Scaling Up

Let's try with 3 qubits (8 states) and 4 qubits (16 states).

```
# 3 qubits = 8 states
n_qubits = 3
qc = QuantumCircuit(n_qubits, n_qubits)

# Apply Hadamard to all qubits
for i in range(n_qubits):
    qc.h(i)

qc.measure_all()

result = simulator.run(qc, shots=1000).result()
print(f"{n_qubits} qubits = {2**n_qubits} states")
plot_histogram(result.get_counts())
```

```
# 4 qubits = 16 states
n_qubits = 4
qc = QuantumCircuit(n_qubits)

for i in range(n_qubits):
    qc.h(i)

qc.measure_all()

result = simulator.run(qc, shots=2000).result()
print(f"{n_qubits} qubits = {2**n_qubits} states")
plot_histogram(result.get_counts())
```

## Exercise 3: Unequal Superpositions

We can create superpositions with unequal probabilities using rotation gates.

```
# RY gate rotates the qubit, changing the probability distribution
qc = QuantumCircuit(1, 1)

# theta = pi/4 gives about 85% |0⟩ and 15% |1⟩
theta = np.pi / 4
qc.ry(theta, 0)
qc.measure(0, 0)

result = simulator.run(qc, shots=1000).result()
print(f"With theta = pi/4:")
plot_histogram(result.get_counts())
```

```
# Try different angles
for angle_fraction in [1/8, 1/4, 3/8, 1/2, 5/8, 3/4]:
    theta = np.pi * angle_fraction
    qc = QuantumCircuit(1, 1)
    qc.ry(theta, 0)
    qc.measure(0, 0)
    
    result = simulator.run(qc, shots=1000).result()
    counts = result.get_counts()
    p0 = counts.get('0', 0) / 1000
    p1 = counts.get('1', 0) / 1000
    print(f"theta = {angle_fraction}*pi: P(0)={p0:.2f}, P(1)={p1:.2f}")
```

## Exercise 4: Superposition is Destroyed by Measurement

Once you measure, the superposition collapses.

```
# Before measurement: superposition exists
qc = QuantumCircuit(1)
qc.h(0)

state = Statevector(qc)
print("State before measurement:")
print(state)
```

```
# Simulate measurement
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

# Each run gives a definite result
for i in range(5):
    result = simulator.run(qc, shots=1).result()
    print(f"Measurement {i+1}: {result.get_counts()}")
```
