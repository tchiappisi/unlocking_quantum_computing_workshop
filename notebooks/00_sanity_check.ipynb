## Step 2: Import Libraries

Import the necessary Qiskit components.

```
# Restart may be needed after install - if you get errors, click Runtime > Restart and run all
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

print("Imports successful!")
```

## Step 3: Create Your First Quantum Circuit

We'll create a simple circuit with:
- 1 qubit
- 1 classical bit (to store the measurement result)
- A Hadamard gate (creates superposition)
- A measurement

```
# Create a quantum circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Add a Hadamard gate to put the qubit in superposition
qc.h(0)

# Measure the qubit
qc.measure(0, 0)

# Draw the circuit with matplotlib
qc.draw('mpl')
```

## Step 4: Run the Circuit on a Simulator

We'll run the circuit 1000 times ("shots") and see the distribution of results.

```
# Create a simulator
simulator = AerSimulator()

# Run the circuit
job = simulator.run(qc, shots=1000)
result = job.result()
counts = result.get_counts()

print("Results:", counts)
```

## Step 5: Visualize the Results

```
# Plot a histogram of the results
plot_histogram(counts)
```
