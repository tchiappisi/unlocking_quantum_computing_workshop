## Part 1: Quantum Random Number Generator

Classical "random" numbers are actually pseudo-random. Quantum randomness is fundamentally unpredictable.

### Task 1.1: Generate a Random Bit

```
def quantum_random_bit():
    """Generate a truly random bit using quantum superposition"""
    qc = QuantumCircuit(1, 1)
    qc.h(0)  # Create superposition
    qc.measure(0, 0)
    
    simulator = AerSimulator()
    result = simulator.run(qc, shots=1).result()
    return int(list(result.get_counts().keys())[0])

# Generate 10 random bits
print("Quantum random bits:")
bits = [quantum_random_bit() for _ in range(10)]
print(bits)
```

### Task 1.2: Generate a Random Number (0-255)

```
def quantum_random_byte():
    """Generate a random 8-bit number (0-255)"""
    qc = QuantumCircuit(8, 8)
    
    # Put all qubits in superposition
    for i in range(8):
        qc.h(i)
    
    # Measure all
    qc.measure(range(8), range(8))
    
    simulator = AerSimulator()
    result = simulator.run(qc, shots=1).result()
    bitstring = list(result.get_counts().keys())[0]
    return int(bitstring, 2)

# Generate some random numbers
print("Quantum random numbers (0-255):")
for _ in range(10):
    print(quantum_random_byte(), end=' ')
```

### Task 1.3: Verify Uniform Distribution

```
# Generate many random numbers and check distribution
qc = QuantumCircuit(4, 4)  # 4 bits = 16 possible values
for i in range(4):
    qc.h(i)
qc.measure(range(4), range(4))

simulator = AerSimulator()
result = simulator.run(qc, shots=10000).result()
counts = result.get_counts()

print(f"Generated {len(counts)} unique values")
print(f"Expected ~625 per value (10000/16)")
print(f"Actual range: {min(counts.values())} - {max(counts.values())}")
plot_histogram(counts)
```

## Part 2: Quantum Coin Flip Game

```
def quantum_coin_flip():
    """Simulate a fair coin flip"""
    qc = QuantumCircuit(1, 1)
    qc.h(0)
    qc.measure(0, 0)
    
    simulator = AerSimulator()
    result = simulator.run(qc, shots=1).result()
    outcome = list(result.get_counts().keys())[0]
    return "Heads" if outcome == '0' else "Tails"

# Play the game
print("Quantum Coin Flip Game!")
print("=" * 30)
heads = 0
tails = 0
for i in range(10):
    result = quantum_coin_flip()
    if result == "Heads":
        heads += 1
    else:
        tails += 1
    print(f"Flip {i+1}: {result}")

print(f"\nFinal: Heads={heads}, Tails={tails}")
```

## Part 3: Quantum State Teleportation

Teleportation uses entanglement to transfer a quantum state without physically moving it.

```
def teleport_state(state_to_send):
    """
    Teleport a quantum state from Alice to Bob.
    state_to_send: 'zero', 'one', or 'superposition'
    """
    # 3 qubits: q0 = state to teleport, q1 = Alice's entangled qubit, q2 = Bob's qubit
    qc = QuantumCircuit(3, 2)
    
    # Step 0: Prepare the state to teleport (on q0)
    if state_to_send == 'one':
        qc.x(0)
    elif state_to_send == 'superposition':
        qc.h(0)
    # else: state is already |0⟩
    
    qc.barrier()
    
    # Step 1: Create entanglement between Alice (q1) and Bob (q2)
    qc.h(1)
    qc.cx(1, 2)
    
    qc.barrier()
    
    # Step 2: Alice performs Bell measurement on q0 and q1
    qc.cx(0, 1)
    qc.h(0)
    
    qc.barrier()
    
    # Step 3: Measure Alice's qubits
    qc.measure(0, 0)
    qc.measure(1, 1)
    
    # Step 4: Bob applies corrections based on Alice's measurements
    # (In real quantum, this would be classical communication)
    # For simulation, we use controlled operations
    qc.cx(1, 2)  # If Alice's q1 was 1, flip Bob's qubit
    qc.cz(0, 2)  # If Alice's q0 was 1, apply Z to Bob's qubit
    
    return qc

# Draw the teleportation circuit
qc = teleport_state('superposition')
qc.draw('mpl')
```

```
# Test teleportation
print("Testing Quantum Teleportation")
print("=" * 40)

simulator = AerSimulator()

for state in ['zero', 'one', 'superposition']:
    qc = teleport_state(state)
    # Add measurement of Bob's qubit
    qc.measure(2, 0)  # Overwrite measurement
    
    result = simulator.run(qc, shots=1000).result()
    print(f"\nSent: |{state}⟩")
    print(f"Bob received: {result.get_counts()}")
```
