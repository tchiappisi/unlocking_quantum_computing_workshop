## Exercise 1: The |0⟩ State

By default, qubits start in the |0⟩ state. Let's verify this.

```
# Create a circuit with 1 qubit - it starts in |0⟩
qc = QuantumCircuit(1, 1)
qc.measure(0, 0)

# Run and see results
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
print("Results:", result.get_counts())
plot_histogram(result.get_counts())
```

**Question:** What do you observe? Why do we get 100% "0"?

## Exercise 2: The |1⟩ State

Use the X gate (NOT gate) to flip the qubit from |0⟩ to |1⟩.

```
qc = QuantumCircuit(1, 1)
qc.x(0)  # Apply X gate to flip |0⟩ to |1⟩
qc.measure(0, 0)

# Draw the circuit
qc.draw('mpl')
```

```
# Run and see results
result = simulator.run(qc, shots=1000).result()
print("Results:", result.get_counts())
plot_histogram(result.get_counts())
```

## Exercise 3: Superposition

Use the Hadamard gate to create a superposition of |0⟩ and |1⟩.

```
qc = QuantumCircuit(1, 1)
qc.h(0)  # Hadamard gate creates superposition
qc.measure(0, 0)

qc.draw('mpl')
```

```
result = simulator.run(qc, shots=1000).result()
print("Results:", result.get_counts())
plot_histogram(result.get_counts())
```

**Question:** Why do we see roughly 50-50 results?

## Exercise 4: Visualize the Qubit State

The Bloch sphere is a way to visualize qubit states.

```
# |0⟩ state - points to North pole
qc = QuantumCircuit(1)
state = Statevector(qc)
plot_bloch_multivector(state)
```

```
# |1⟩ state - points to South pole
qc = QuantumCircuit(1)
qc.x(0)
state = Statevector(qc)
plot_bloch_multivector(state)
```

```
# Superposition - points to equator
qc = QuantumCircuit(1)
qc.h(0)
state = Statevector(qc)
plot_bloch_multivector(state)
```
